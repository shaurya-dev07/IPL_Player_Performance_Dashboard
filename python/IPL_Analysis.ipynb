5. Data Cleaning and Standardization.

```python
# Create copies of original datasets
matches_clean = matches.copy()
deliveries_clean = deliveries.copy()

# Remove leading/trailing spaces from text columns
for df in [matches_clean, deliveries_clean]:
    for col in df.columns:
        if pd.api.types.is_string_dtype(df[col]):
            df[col] = df[col].str.strip()

# Convert date into datetime format
matches_clean["date"] = pd.to_datetime(
    matches_clean["date"], errors="coerce"
)

# Keep season as text
matches_clean["season"] = matches_clean["season"].astype("string")

print("First-pass cleaning completed!")
```

```python
team_mapping = {
    "Delhi Daredevils": "Delhi Capitals",
    "Kings XI Punjab": "Punjab Kings",
    "Royal Challengers Bangalore": "Royal Challengers Bengaluru",
    "Rising Pune Supergiants": "Rising Pune Supergiant"
}

# Standardize teams in matches
match_team_columns = [
    "team1", "team2", "toss_winner", "winner"
]

for col in match_team_columns:
    matches_clean[col] = matches_clean[col].replace(team_mapping)

# Standardize teams in deliveries
delivery_team_columns = [
    "batting_team", "bowling_team"
]

for col in delivery_team_columns:
    deliveries_clean[col] = deliveries_clean[col].replace(team_mapping)

print("Team names standardized!")
```

```python
print("MATCHES:", matches_clean.shape)
print("DELIVERIES:", deliveries_clean.shape)

print("\nMissing values after cleaning:")
print("Matches:", matches_clean.isnull().sum().sum())
print("Deliveries:", deliveries_clean.isnull().sum().sum())

print("\nStandardized teams:")
teams = pd.unique(pd.concat([
    matches_clean["team1"],
    matches_clean["team2"]
]).dropna())

print(sorted(teams))
```

6. Player Performance KPI Calculations

```python
# Create a unique identifier for each innings
deliveries_clean["innings_id"] = (
    deliveries_clean["match_id"].astype(str)
    + "_"
    + deliveries_clean["inning"].astype(str)
)

# Calculate batting statistics
batting_stats = deliveries_clean.groupby("batter").agg(
    innings=("innings_id", "nunique"),
    runs=("batsman_runs", "sum"),
    balls_faced=("extras_type", lambda x: (x != "wides").sum())
)

# Calculate dismissals (retired hurt is not counted as out)
dismissals = deliveries_clean[
    deliveries_clean["player_dismissed"].notna()
    & (deliveries_clean["dismissal_kind"] != "retired hurt")
].groupby("player_dismissed").size()

batting_stats["dismissals"] = dismissals.reindex(
    batting_stats.index, fill_value=0
)

# Calculate batting average and strike rate
batting_stats["batting_average"] = (
    batting_stats["runs"] /
    batting_stats["dismissals"].replace(0, np.nan)
)

batting_stats["strike_rate"] = (
    batting_stats["runs"] /
    batting_stats["balls_faced"].replace(0, np.nan)
) * 100

batting_stats = batting_stats.round(2)

# Sort by total runs
batting_stats = batting_stats.sort_values(
    "runs", ascending=False
)

print("TOP 10 BATSMEN")
display(batting_stats.head(10))
```

```python
# Wickets credited to the bowler
bowler_wickets = [
    "bowled",
    "caught",
    "caught and bowled",
    "lbw",
    "stumped",
    "hit wicket"
]

# Exclude byes, legbyes and penalty runs from bowler's conceded runs
deliveries_clean["bowler_runs"] = (
    deliveries_clean["total_runs"]
    - deliveries_clean["extras_type"].isin(
        ["byes", "legbyes", "penalty"]
    ) * deliveries_clean["extra_runs"]
)

# Legal balls exclude wides and no-balls
deliveries_clean["legal_ball"] = (
    ~deliveries_clean["extras_type"].isin(["wides", "noballs"])
)

# Calculate bowling statistics
bowling_stats = deliveries_clean.groupby("bowler").agg(
    innings=("innings_id", "nunique"),
    runs_conceded=("bowler_runs", "sum"),
    balls_bowled=("legal_ball", "sum")
)

# Calculate wickets credited to bowlers
wickets = deliveries_clean[
    deliveries_clean["dismissal_kind"].isin(bowler_wickets)
].groupby("bowler").size()

bowling_stats["wickets"] = wickets.reindex(
    bowling_stats.index, fill_value=0
)

# Calculate bowling economy, average and strike rate
bowling_stats["overs"] = (
    bowling_stats["balls_bowled"] // 6
    + (bowling_stats["balls_bowled"] % 6) / 10
)

bowling_stats["economy"] = (
    bowling_stats["runs_conceded"] * 6
    / bowling_stats["balls_bowled"].replace(0, np.nan)
)

bowling_stats["bowling_average"] = (
    bowling_stats["runs_conceded"]
    / bowling_stats["wickets"].replace(0, np.nan)
)

bowling_stats["bowling_strike_rate"] = (
    bowling_stats["balls_bowled"]
    / bowling_stats["wickets"].replace(0, np.nan)
)

bowling_stats = bowling_stats.round(2)

bowling_stats = bowling_stats.sort_values(
    "wickets", ascending=False
)

print("TOP 10 BOWLERS")
display(bowling_stats.head(10))
```

```python
# Create match ID and season mapping
match_seasons = matches_clean[["id", "season"]].copy()

# Merge season into deliveries
season_deliveries = deliveries_clean.merge(
    match_seasons,
    left_on="match_id",
    right_on="id",
    how="left"
)

print("Season mapping completed!")
print("Rows:", len(season_deliveries))
print("Missing seasons:", season_deliveries["season"].isnull().sum())
```

```python
season_batting = season_deliveries.groupby(
    ["season", "batter"]
).agg(
    runs=("batsman_runs", "sum"),
    balls_faced=("extras_type", lambda x: (x != "wides").sum()),
    innings=("innings_id", "nunique")
).reset_index()

season_dismissals = season_deliveries[
    season_deliveries["player_dismissed"].notna()
    & (season_deliveries["dismissal_kind"] != "retired hurt")
].groupby(
    ["season", "player_dismissed"]
).size().reset_index(name="dismissals")

season_batting = season_batting.merge(
    season_dismissals,
    left_on=["season", "batter"],
    right_on=["season", "player_dismissed"],
    how="left"
)

season_batting["dismissals"] = (
    season_batting["dismissals"].fillna(0)
)

season_batting["average"] = (
    season_batting["runs"] /
    season_batting["dismissals"].replace(0, np.nan)
)

season_batting["strike_rate"] = (
    season_batting["runs"] /
    season_batting["balls_faced"].replace(0, np.nan)
) * 100

season_batting = season_batting.drop(
    columns=["player_dismissed"]
).round(2)

print("TOP 10 RUN SCORERS BY SEASON")

display(
    season_batting.sort_values(
        ["season", "runs"],
        ascending=[True, False]
    ).groupby("season").head(10)
)
```

```python
# Calculate season-wise bowling statistics
season_bowling = season_deliveries.groupby(
    ["season", "bowler"]
).agg(
    innings=("innings_id", "nunique"),
    runs_conceded=("bowler_runs", "sum"),
    balls_bowled=("legal_ball", "sum")
).reset_index()

# Calculate wickets credited to bowlers
season_wickets = season_deliveries[
    season_deliveries["dismissal_kind"].isin(bowler_wickets)
].groupby(
    ["season", "bowler"]
).size().reset_index(name="wickets")

# Merge wickets
season_bowling = season_bowling.merge(
    season_wickets,
    on=["season", "bowler"],
    how="left"
)

season_bowling["wickets"] = (
    season_bowling["wickets"].fillna(0).astype(int)
)

# Calculate bowling overs
season_bowling["overs"] = (
    season_bowling["balls_bowled"] // 6
    + (season_bowling["balls_bowled"] % 6) / 10
)

# Calculate bowling economy
season_bowling["economy"] = (
    season_bowling["runs_conceded"] * 6
    / season_bowling["balls_bowled"].replace(0, np.nan)
)

# Calculate bowling average
season_bowling["bowling_average"] = (
    season_bowling["runs_conceded"]
    / season_bowling["wickets"].replace(0, np.nan)
)

# Calculate bowling strike rate
season_bowling["bowling_strike_rate"] = (
    season_bowling["balls_bowled"]
    / season_bowling["wickets"].replace(0, np.nan)
)

season_bowling = season_bowling.round(2)

print("TOP 10 WICKET-TAKERS BY SEASON")

display(
    season_bowling.sort_values(
        ["season", "wickets"],
        ascending=[True, False]
    ).groupby("season").head(10)
)
```
